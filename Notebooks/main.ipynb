# Dataset: Yellow Taxi Trip Records - March

```python
import pyarrow
import pandas as pd
import numpy as np
from pathlib import Path
```

## 1.Extract - Data Read

```python
df = pd.read_csv('/kaggle/input/datasets/elemento/nyc-yellow-taxi-trip-data/yellow_tripdata_2016-03.csv')
print(f"Total de registros: {len(df):,}")
print(f"Total de colunas: {len(df.columns)}")
df.head()
```

## 2. Data Quality & Anomalies
### Detect and remove invalid registers

```python
# Convert column data
df['pickup_datetime'] = pd.to_datetime(df['tpep_pickup_datetime'])
df['dropoff_datetime'] = pd.to_datetime(df['tpep_dropoff_datetime'])

# Check for null values

print("====Null Values====")
null_counts = df.isnull().sum()
#print(null_counts)
print(null_counts[null_counts > 0] if null_counts.sum() > 0 else "No null values found")

```

```python
# Detecting Anomalies
total = len(df)

# Invalid Distance – compares the race distance; ex: it cannot be 0 miles or more than 100 miles.
invalid_distance = (df['trip_distance'] <= 0) | (df['trip_distance'] > 100)

# Invalid number of passengers
invalid_passengers = (df['passenger_count'] <= 0) | (df['passenger_count'] > 6)

# Negative fare
invalid_fare = df['fare_amount'] < 0

# Invalid duration (pickup after drop-off)
invalid_duration = df['pickup_datetime'] > df['dropoff_datetime']

# Coordinates outside NYC (latitude ~40.4-41.0, longitude ~-74.3 a -73.7)
invalid_pickup_coords = (
    (df['pickup_latitude'] < 40.4) | (df['pickup_latitude'] > 41.0) |
    (df['pickup_longitude'] < -74.3) | (df['pickup_longitude'] > -73.7)
)
invalid_dropoff_coords = (
    (df['dropoff_latitude'] < 40.4) | (df['dropoff_latitude'] > 41.0) |
    (df['dropoff_longitude'] < -74.3) | (df['dropoff_longitude'] > -73.7)
)

print("=== ANOMALIES DETECTED ===")
print(f"Invalid distance (<=0 ou >100mi): {invalid_distance.sum():,} ({invalid_distance.sum()/total*100:.2f}%)")
print(f"Invalid passengers (<=0 ou >6):  {invalid_passengers.sum():,} ({invalid_passengers.sum()/total*100:.7f}%)")
print(f"Negative tariff:                    {invalid_fare.sum():,} ({invalid_fare.sum()/total*100:.4f}%)")
print(f"Invalid duration:                   {invalid_duration.sum():,} ({invalid_duration.sum()/total*100:.7f}%)")
print(f"Pickup coordinates outside NYC:     {invalid_pickup_coords.sum():,} ({invalid_pickup_coords.sum()/total*100:.2f}%)")
print(f"Drop-off coordinates outside NYC:   {invalid_dropoff_coords.sum():,} ({invalid_dropoff_coords.sum()/total*100:.2f}%)")
```

```python
# Remove Invalid Records
invalid_mask = (
   invalid_distance |
   invalid_passengers |
   invalid_fare |
   invalid_duration |
   invalid_pickup_coords |
   invalid_dropoff_coords 
)

df_clean = df[~invalid_mask].copy()
removed = total - len(df_clean)
print(f"Previous records:   {total:,}")
print(f"Removed records: {removed:,} ({removed/total*100:.2f}%)")
print(f"Records after:    {len(df_clean):,}")
```

## 3. Transform - Calculated Columns
### Enrich the dataset with derived metrics

```python
# Run duration in minutes
df_clean['trip_duration_min'] = (
    (df_clean['dropoff_datetime'] - df_clean['pickup_datetime'])
    .dt.total_seconds()/60
)

# Average speed (mph)
hours = df_clean['trip_duration_min']/60
df_clean['trip_speed_mph'] = (df_clean['trip_distance']/hours.replace(0,np.nan)).fillna(0)

# Revenue per mile
df_clean['revenue_per_mile'] = (df_clean['total_amount']/df_clean['trip_distance'].replace(0, np.nan)).fillna(0)

# Hour of day
df_clean['hour_of_day'] = df_clean['pickup_datetime'].dt.hour

# Date (without hour)
df_clean['date'] = df_clean['pickup_datetime'].dt.date

# Day of week
df_clean['day_of_week'] = df_clean['pickup_datetime'].dt.day_name()

# Percentual Tip
df_clean ['tip_pct'] = ( df_clean['tip_amount']/df_clean['fare_amount'].replace(0,np.nan)*100).fillna(0)

print("=== NEW COLUMNS ===")
new_cols = ['trip_duration_min', 'trip_speed_mph', 'revenue_per_mile', 'hour_of_day', 'date', 'day_of_week', 'tip_pct']
for col in new_cols:
    print(f"  {col}")

print(f"\nTotal of Columns: {len(df_clean.columns)}")
df_clean[new_cols].head(10)
```

## 4. Aggregations and Metrics¶
### Summaries by hour, vendor, payment type, and zone

```python
# Summary per hour

hourly = df_clean.groupby('hour_of_day').agg(
    total_trips = ('hour_of_day', 'size'),
    avg_distance = ('trip_distance', 'mean'),
    avg_fare = ('fare_amount', 'mean'),
    avg_duration = ('trip_duration_min','mean'),
    avg_speed = ('trip_speed_mph','mean'),
    avg_tip_pct = ('tip_pct','mean'),
 ).round(2)

print('===SUMMARY PER HOUR PER DAY===')
hourly
```

```python
# Summary per vendor

vendor = df_clean.groupby('VendorID').agg(
    total_trips = ('VendorID', 'size'),
    avg_distance = ('trip_distance', 'mean'),
    avg_fare = ('fare_amount','mean'),
    total_revenue = ('total_amount','sum'),
    avg_tip_pct = ('tip_pct','mean'),
    avg_speed = ('trip_speed_mph', 'mean'),
).round(2)

print("=== SUMMARY PER VENDOR ===")
vendor
```

```python
# Summary per type of payment

payment_labels = {1: 'Credit Card', 2: 'Cash', 3: 'No Charge', 4: 'Dispute', 5: 'Unknown', 6: 'Voided'}
df_clean['payment_label'] = df_clean['payment_type'].map(payment_labels).fillna('Other')

payment = df_clean.groupby('payment_label').agg(
    total_trips = ('payment_label','size'),
    avg_fare = ('fare_amount', 'mean'),
    avg_tip = ('tip_amount', 'mean'),
    avg_tip_pct = ('tip_pct', 'mean'),
    total_revenue = ('total_amount', 'sum'),
    
).round(2)

payment['pct_trips'] = ((payment['total_trips'] / payment['total_trips'].sum()) * 100).round(2)

print("=== SUMMARY BY PAYMENT TYPE ===")
payment
```

```python
# Summary per day of week

weekday = df_clean.groupby('day_of_week').agg(
    total_trips = ('day_of_week','size'),
    avg_distance = ('trip_distance','mean'),
    avg_fare = ('fare_amount','mean'),
    avg_duration = ('trip_duration_min', 'mean'),
    avg_tip_pct = ('tip_pct', 'mean'),
).round(2)

day_order = ['Monday', 'Tuesday', 'Wednesday', 'Thursday', 'Friday', 'Saturday', 'Sunday']
weekday = weekday.reindex(day_order)

print("=== SUMMARY BY DAY OF THE WEEK ===")
weekday
```

## 5. Load - Parquet
### Salvar os dados limpos em formato Parquet

```python
# Convert date column to string (pyarrow does not serialize datetime.date)
df_clean['date'] = df_clean['date'].astype(str)

# Remove original datetime columns (we already have pickup_datetime and dropoff_datetime)
df_clean = df_clean.drop(columns=['tpep_pickup_datetime', 'tpep_dropoff_datetime'])

output_path = Path('/kaggle/working/output/yellow_taxi_2016-03.parquet')
output_path.parent.mkdir(parents=True, exist_ok=True)

df_clean.to_parquet(output_path, engine='pyarrow', index=False)

size_mb = output_path.stat().st_size / (1024 * 1024)
print(f"=== PARQUET ===")
print(f"File: {output_path}")
print(f"Size: {size_mb:.1f} MB")
```

```python
# Validation: read back the Parquet and verify
df_parquet = pd.read_parquet(output_path)
print(f"=== VALIDATION ===")
print(f"Rcords on Parquet: {len(df_parquet):,}")
print(f"Expected Records:  {len(df_clean):,}")
print(f"Match: {'OK' if len(df_parquet) == len(df_clean) else 'ERRO'}")
print(f"\nColunas: {list(df_parquet.columns)}")
df_parquet.head()
```
